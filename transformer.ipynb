```python
n_ctx = 1024
n_embd = 768
n_vocab = 50257
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
import tiktoken
```

```python
class Embeddings(nn.Module):
    def __init__(self, n_ctx, n_vocab, n_embd):
        super().__init__()
        self.w_embedding = nn.Embedding(n_vocab, n_embd)
        self.pos_embedding = nn.Embedding(n_ctx, n_embd)
    def forward(self, x):
        x = self.w_embedding(x) + self.pos_embedding(torch.arange(x.shape[-1], device=x.device))
        return x
```

```python
from datasets import load_dataset
```

```python
ds = load_dataset("Hananguyen12/Shakespeare-works", split="train")
text = "\n".join(ds["text"])
```

```python
enc = tiktoken.get_encoding("gpt2")
tokens = torch.tensor(enc.encode(text), dtype=torch.long)
batch_size = 4
seq_len = 64

# Grab 4 random starting positions in the text and stack them into a (4, 64) batch
ix = torch.randint(len(tokens) - seq_len, (batch_size,))
x = torch.stack([tokens[i : i + seq_len] for i in ix])
```

```python
embed = Embeddings(n_ctx, n_vocab, n_embd)
out = embed(x)
out.shape
```
